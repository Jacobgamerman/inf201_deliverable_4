```python
"""
Task 1
"""


class Temperature:
    """Represents a temperature in Celsius, Fahrenheit, or Kelvin."""

    def __init__(self, value: float, unit: str = "C") -> None:
        """Initializes a Temperature instance with a value and unit.

        Args:
            value: The numerical value of the temperature.
            unit: The temperature unit ('C', 'F', or 'K'), case-insensitive.
                Defaults to 'C'.

        Raises:
            ValueError: If `unit` is invalid or temperature is below absolute zero.
        """
        self._value: float = float(value)
        if not isinstance(unit, str):
            raise ValueError("Unit must be string")
        self._unit: str = unit.upper()
        self.legalunits: tuple[str, ...] = ("C", "F", "K")
        self.islegalunit()
        self.subzero()

    def to_celsius(self) -> float:
        """Converts the temperature to Celsius.

        Returns:
            The temperature in Celsius.
        """
        if self._unit == "K":
            return self._value - 273.15

        elif self._unit == "F":
            return (self._value - 32) * (5 / 9)

        return self._value

    def to_fahrenheit(self) -> float:
        """Converts the temperature to Fahrenheit.

        Returns:
            The temperature in Fahrenheit.
        """
        if self._unit == "C":
            return self._value * (9 / 5) + 32
        if self._unit == "K":
            return (self._value - 273.15) * (9 / 5) + 32

        return self._value

    def to_kelvin(self) -> float:
        """Converts the temperature to Kelvin.

        Returns:
            The temperature in Kelvin.
        """
        if self._unit == "C":
            return self._value + 273.15
        elif self._unit == "F":
            return (self._value - 32) * (5 / 9) + 273.15

        return self._value

    def convert_to(self, target_unit: str) -> "Temperature":
        """Returns a new Temperature instance converted to target_unit.

        Args:
            target_unit: The unit to convert to ('C', 'F', or 'K').

        Returns:
            A new Temperature instance in the target unit.

        Raises:
            ValueError: If `target_unit` is not an allowed unit.
        """
        if not isinstance(target_unit, str) or target_unit.upper() not in self.legalunits:
            raise ValueError("The target unit given is not allowed")
        target: str = target_unit.upper()

        if target == "C":
            return Temperature(self.to_celsius(), "C")
        elif target == "K":
            return Temperature(self.to_kelvin(), "K")
        return Temperature(self.to_fahrenheit(), "F")

    def __repr__(self) -> str:
        """Returns an unambiguous string representation of the Temperature."""
        return f"Temperature({self._value}, '{self._unit}')"

    def __str__(self) -> str:
        """Returns a human-readable string formatted to two decimal places."""
        return f"{self._value:.2f} °{self._unit}"

    def subzero(self) -> None:
        """Checks that the temperature is not below absolute zero (0 K)."""
        if self.to_kelvin() < -1e-9:
            raise ValueError("Temperatures under zero kelvin is not valid")
        else:
            pass

    def islegalunit(self) -> None:
        """Checks that the unit is one of the allowed temperature units."""
        if self.unit in self.legalunits:
            pass
        else:
            raise ValueError("The unit given is not allowed")

    @property
    def unit(self) -> str:
        """str: The read-only temperature unit."""
        return self._unit

    @property
    def value(self) -> float:
        """float: The read-only temperature value."""
        return self._value

    def __eq__(self, other: object) -> bool:
        """Checks if two temperatures are equal within 1e-9 K tolerance."""
        if not isinstance(other, Temperature):
            return NotImplemented
        return abs(self.to_kelvin() - other.to_kelvin()) < 1e-9

    def __ne__(self, other: object) -> bool:
        """Checks if two temperatures are not equal within 1e-9 K tolerance."""
        if not isinstance(other, Temperature):
            return NotImplemented
        return not (self == other)

    def __lt__(self, other: object) -> bool:
        """Checks if this temperature is less than another."""
        if not isinstance(other, Temperature):
            return NotImplemented
        return (self.to_kelvin() < other.to_kelvin() - 1e-9)

    def __le__(self, other: object) -> bool:
        """Checks if this temperature is less than or equal to another."""
        if not isinstance(other, Temperature):
            return NotImplemented
        return (self.to_kelvin() <= other.to_kelvin() + 1e-9)

    def __gt__(self, other: object) -> bool:
        """Checks if this temperature is greater than another."""
        if not isinstance(other, Temperature):
            return NotImplemented
        return (self.to_kelvin() > other.to_kelvin() + 1e-9)

    def __ge__(self, other: object) -> bool:
        """Checks if this temperature is greater than or equal to another."""
        if not isinstance(other, Temperature):
            return NotImplemented
        return (self.to_kelvin() >= other.to_kelvin() - 1e-9)
```

```python
"""
Task 2
"""


import math


class Fraction:
    """Represents a mathematical fraction with an integer numerator and a strictly positive integer denominator."""

    def __init__(self, numerator: int, denominator: int = 1) -> None:
        """Initializes and simplifies a Fraction instance."""
        if not isinstance(numerator, int) or isinstance(numerator, bool):
            raise TypeError("Numerator must be an integer.")
        if not isinstance(denominator, int) or isinstance(denominator, bool):
            raise TypeError("Denominator must be an integer.")
        if denominator <= 0:
            raise ValueError("Denominator must be strictly positive (> 0).")

        common_divisor = math.gcd(numerator, denominator)
        self.numerator: int = numerator // common_divisor
        self.denominator: int = denominator // common_divisor

    # --- String representations (Extra magic methods 1 & 2) ---

    def __str__(self) -> str:
        """Returns a readable string representation of the fraction."""
        if self.denominator == 1:
            return f"{self.numerator}"
        return f"{self.numerator}/{self.denominator}"

    def __repr__(self) -> str:
        """Returns an unambiguous string representation for debugging."""
        return f"Fraction({self.numerator}, {self.denominator})"

    # --- Comparison methods (Required) ---

    def __eq__(self, other: object) -> bool:
        """Checks if two fractions are equal (==)."""
        if not isinstance(other, Fraction):
            return NotImplemented
        return self.numerator == other.numerator and self.denominator == other.denominator

    def __ne__(self, other: object) -> bool:
        """Checks if two fractions are not equal (!=)."""
        if not isinstance(other, Fraction):
            return NotImplemented
        return not self.__eq__(other)

    def __lt__(self, other: "Fraction") -> bool:
        """Checks if this fraction is strictly smaller than another (<)."""
        if not isinstance(other, Fraction):
            return NotImplemented
        return self.numerator * other.denominator < other.numerator * self.denominator

    def __le__(self, other: "Fraction") -> bool:
        """Checks if this fraction is smaller than or equal to another (<=)."""
        if not isinstance(other, Fraction):
            return NotImplemented
        return self.numerator * other.denominator <= other.numerator * self.denominator

    def __gt__(self, other: "Fraction") -> bool:
        """Checks if this fraction is strictly greater than another (>)."""
        if not isinstance(other, Fraction):
            return NotImplemented
        return self.numerator * other.denominator > other.numerator * self.denominator

    def __ge__(self, other: "Fraction") -> bool:
        """Checks if this fraction is greater than or equal to another (>=)."""
        if not isinstance(other, Fraction):
            return NotImplemented
        return self.numerator * other.denominator >= other.numerator * self.denominator

    # --- Arithmetic operations (Extra magic methods 3, 4, 5 & 6) ---

    def __add__(self, other: "Fraction") -> "Fraction":
        """Adds two fractions (+)."""
        if not isinstance(other, Fraction):
            return NotImplemented
        new_num = self.numerator * other.denominator + other.numerator * self.denominator
        new_den = self.denominator * other.denominator
        return Fraction(new_num, new_den)

    def __sub__(self, other: "Fraction") -> "Fraction":
        """Subtracts another fraction from this fraction (-)."""
        if not isinstance(other, Fraction):
            return NotImplemented
        new_num = self.numerator * other.denominator - other.numerator * self.denominator
        new_den = self.denominator * other.denominator
        return Fraction(new_num, new_den)

    def __mul__(self, other: "Fraction") -> "Fraction":
        """Multiplies two fractions (*)."""
        if not isinstance(other, Fraction):
            return NotImplemented
        return Fraction(self.numerator * other.numerator, self.denominator * other.denominator)

    def __truediv__(self, other: "Fraction") -> "Fraction":
        """Divides this fraction by another fraction (/)."""
        if not isinstance(other, Fraction):
            return NotImplemented
        if other.numerator == 0:
            raise ZeroDivisionError("Cannot divide by a zero fraction.")
        if other.numerator < 0:
            return Fraction(-self.numerator * other.denominator, self.denominator * abs(other.numerator))
        return Fraction(self.numerator * other.denominator, self.denominator * other.numerator)
```
