```python
"""
Task 2
"""


import math


class Fraction:
    """Represents a mathematical fraction with an integer numerator and a strictly positive integer denominator."""

    def __init__(self, numerator: int, denominator: int = 1) -> None:
        """Initializes and simplifies a Fraction instance."""
        if not isinstance(numerator, int) or isinstance(numerator, bool):
            raise TypeError("Numerator must be an integer.")
        if not isinstance(denominator, int) or isinstance(denominator, bool):
            raise TypeError("Denominator must be an integer.")
        if denominator <= 0:
            raise ValueError("Denominator must be strictly positive (> 0).")

        common_divisor = math.gcd(numerator, denominator)
        self.numerator: int = numerator // common_divisor
        self.denominator: int = denominator // common_divisor

    # --- String representations (Extra magic methods 1 & 2) ---

    def __str__(self) -> str:
        """Returns a readable string representation of the fraction."""
        if self.denominator == 1:
            return f"{self.numerator}"
        return f"{self.numerator}/{self.denominator}"

    def __repr__(self) -> str:
        """Returns an unambiguous string representation for debugging."""
        return f"Fraction({self.numerator}, {self.denominator})"

    # --- Comparison methods (Required) ---

    def __eq__(self, other: object) -> bool:
        """Checks if two fractions are equal (==)."""
        if not isinstance(other, Fraction):
            return NotImplemented
        return self.numerator == other.numerator and self.denominator == other.denominator

    def __ne__(self, other: object) -> bool:
        """Checks if two fractions are not equal (!=)."""
        if not isinstance(other, Fraction):
            return NotImplemented
        return not self.__eq__(other)

    def __lt__(self, other: "Fraction") -> bool:
        """Checks if this fraction is strictly smaller than another (<)."""
        if not isinstance(other, Fraction):
            return NotImplemented
        return self.numerator * other.denominator < other.numerator * self.denominator

    def __le__(self, other: "Fraction") -> bool:
        """Checks if this fraction is smaller than or equal to another (<=)."""
        if not isinstance(other, Fraction):
            return NotImplemented
        return self.numerator * other.denominator <= other.numerator * self.denominator

    def __gt__(self, other: "Fraction") -> bool:
        """Checks if this fraction is strictly greater than another (>)."""
        if not isinstance(other, Fraction):
            return NotImplemented
        return self.numerator * other.denominator > other.numerator * self.denominator

    def __ge__(self, other: "Fraction") -> bool:
        """Checks if this fraction is greater than or equal to another (>=)."""
        if not isinstance(other, Fraction):
            return NotImplemented
        return self.numerator * other.denominator >= other.numerator * self.denominator

    # --- Arithmetic operations (Extra magic methods 3, 4, 5 & 6) ---

    def __add__(self, other: "Fraction") -> "Fraction":
        """Adds two fractions (+)."""
        if not isinstance(other, Fraction):
            return NotImplemented
        new_num = self.numerator * other.denominator + other.numerator * self.denominator
        new_den = self.denominator * other.denominator
        return Fraction(new_num, new_den)

    def __sub__(self, other: "Fraction") -> "Fraction":
        """Subtracts another fraction from this fraction (-)."""
        if not isinstance(other, Fraction):
            return NotImplemented
        new_num = self.numerator * other.denominator - other.numerator * self.denominator
        new_den = self.denominator * other.denominator
        return Fraction(new_num, new_den)

    def __mul__(self, other: "Fraction") -> "Fraction":
        """Multiplies two fractions (*)."""
        if not isinstance(other, Fraction):
            return NotImplemented
        return Fraction(self.numerator * other.numerator, self.denominator * other.denominator)

    def __truediv__(self, other: "Fraction") -> "Fraction":
        """Divides this fraction by another fraction (/)."""
        if not isinstance(other, Fraction):
            return NotImplemented
        if other.numerator == 0:
            raise ZeroDivisionError("Cannot divide by a zero fraction.")
        if other.numerator < 0:
            return Fraction(-self.numerator * other.denominator, self.denominator * abs(other.numerator))
        return Fraction(self.numerator * other.denominator, self.denominator * other.numerator)
```

## Implementation & Design Discussion: Fraction Class

### 1. Object-Oriented Design & Initialization
The `Fraction` class represents rational numbers exactly, avoiding floating-point precision errors. During initialization (`__init__`), strict input validation is enforced:
* **Type Validation**: Uses `isinstance` to ensure both the numerator and denominator are integers, raising a `TypeError` if invalid types are passed.
* **Value Validation**: Checks that the denominator is strictly positive, raising a `ValueError` for zero or negative denominators.
* **Automatic Simplification**: Utilizes Python's built-in `math.gcd` function to compute the greatest common divisor and automatically reduce every fraction to its simplest form upon creation.

### 2. String Representations & Comparisons
* **Formatting**: Implemented `__str__` for a clean, human-readable display (e.g., `'3/4'`, or just the numerator if the denominator is `1`) and `__repr__` for formal debugging representation.
* **Rich Comparisons**: Overloaded all six comparison operators (`==`, `!=`, `<`, `<=`, `>`, `>=`). To prevent floating-point rounding inaccuracies, comparisons are performed using exact integer cross-multiplication.

### 3. Arithmetic Operations
The class overloads standard arithmetic operators (`__add__`, `__sub__`, `__mul__`, `__truediv__`). Each method executes exact rational arithmetic and returns a new, automatically simplified `Fraction` instance. Division includes safety checks to raise a `ZeroDivisionError` when attempting to divide by a fraction with a zero numerator.

### 4. Unit Testing & Verification
A robust `pytest` suite (`test_fraction.py`) validates the implementation across multiple test cases:
* Verifying initialization and automatic fraction reduction.
* Testing exception handling (`TypeError` and `ValueError`) using `pytest.raises`.
* Checking exact `str` and `repr` string formatting.
* Validating all six comparison operators with positive and negative inputs.
* Ensuring correctness in arithmetic calculations and edge cases like division by zero.
